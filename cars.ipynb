```python
# 1. Gestion de la flotte automobile 

class Vehicle:
    def __init__(self, vid, marque, modele, categorie, tarif, etat):
        self.id = vid
        self.marque = marque
        self.modele = modele
        self.categorie = categorie
        self.tarif = tarif
        self.etat = etat  # ex. "disponible", "loué", "entretien"

    def __str__(self):
        return f"{self.marque} {self.modele} ({self.categorie})"


class Car(Vehicle):
    def __init__(self, vid, marque, modele, categorie, tarif, etat):
        super().__init__(vid, marque, modele, categorie, tarif, etat) #appel classe parente depuis enfant


class Truck(Vehicle):
    def __init__(self, vid, marque, modele, categorie, tarif, etat):
        super().__init__(vid, marque, modele, categorie, tarif, etat)


class Motorcycle(Vehicle):
    def __init__(self, vid, marque, modele, categorie, tarif, etat):
        super().__init__(vid, marque, modele, categorie, tarif, etat)

# 4. Classe centrale CarRentalSystem

class CarRentalSystem:
    def __init__(self):
        self.vehicles = []
        self.customers = []
        self.rentals = []

    # Gestion véhicules

    def add_vehicle(self, vehicle):
        self.vehicles.append(vehicle)

    def find_vehicle_by_id(self, vid):
        for v in self.vehicles:
            if v.id == vid:
                return v
        return None

    def available_vehicles(self):
        disponibles = []
        for v in self.vehicles:
            if v.etat == "disponible":
                disponibles.append(v)
        return disponibles

    # Gestion clients

    def add_customer(self, customer):
        self.customers.append(customer)

    def find_customer_by_id(self, cid):
        for c in self.customers:
            if c.id == cid:
                return c
        return None


    # Gestion locations

    def add_rental(self, rental):
        if rental.vehicle.etat != "disponible":
            raise ValueError("Véhicule non disponible")
        rental.vehicle.etat = "loué"
        self.rentals.append(rental)

    def rentals_in_progress(self):
        en_cours = []
        for r in self.rentals:
            if r.en_cours():
                en_cours.append(r)
        return en_cours


    #Rapports
    def report_available_vehicles(self):
        lignes = []
        for v in self.available_vehicles():
            lignes.append(str(v))
        return lignes

    def report_rentals(self):
        lignes = []
        for r in self.rentals:
            lignes.append(str(r))
        return lignes

    def revenue(self):
        total = 0
        for r in self.rentals:
            total += r.cout_total
        return total


# TESTS

system = CarRentalSystem()

v1 = Car(1, "Bugatti", "Chiron Super Sport", "Hypercars", 3000, "disponible")
v2 = Car(2, "Koenigsegg", "Jesko Absolut", "Hypercars", 3500, "disponible")
v3 = Car(3, "Pagani", "Huayra BC", "Hypercars", 2800, "disponible")
v4 = Car(4, "McLaren", "Speedtail", "Hypercars", 2700, "loué")
v5 = Car(5, "Lamborghini", "Sián FKP 37", "Hypercars", 2600, "disponible")
v6 = Truck(2, "Mercedes", "Actros", "Camion", 120, "disponible")
v7 = Motorcycle(3, "Yamaha", "MT-07", "Moto", 60, "loué")

system.add_vehicle(v1)
system.add_vehicle(v2)
system.add_vehicle(v3)
system.add_vehicle(v4)
system.add_vehicle(v5)
system.add_vehicle(v6)
system.add_vehicle(v7)

print(system.find_vehicle_by_id(2)) # recherche de la caisse avec id=2

#afficher les caisses dispo
for veh in system.available_vehicles():
    print(veh)

print(system.report_available_vehicles())
```
